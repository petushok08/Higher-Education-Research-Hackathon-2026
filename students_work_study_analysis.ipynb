# МОЯ ЗАДАЧА

**Совмещение учёбы и работы: польза и трудности для студентов**

---

**Контекст**

Многие студенты начинают работать ещё во время учёбы в университете. Работа может давать опыт, деньги и понимание профессии, но может и отнимать время у учёбы, сна и отдыха. Я хочу разобраться на реальных данных, в чём совмещение учёбы и работы помогает студентам, а в чём мешает.

Задача поставлена в рамках Хакатона исследователей высшего образования – 2026 (НИУ ВШЭ).

---

**Цель проекта:**
Выяснить, в каких аспектах совмещение учёбы и работы приносит студентам пользу, а в каких создаёт дополнительные трудности. Для этого нужно оценить, как с образовательным опытом, успеваемостью и здоровьем студентов связаны:
- сам факт работы;
- **интенсивность** занятости — полный день, неполный день, нерегулярные приработки, количество часов в неделю;
- **тип** занятости — сфера работы, соответствие специальности, мотив, по которому студент пошёл работать.

---

**Ключевые вопросы для анализа**
- Сколько студентов работают во время учёбы, сколько часов, в каких сферах и зачем?
- Отличаются ли работающие и неработающие студенты по **успеваемости** (оценки за сессию)?
- Связана ли работа с **образовательным опытом**: удовлетворённостью обучением, мыслями об отчислении, готовностью снова выбрать тот же вуз и специальность?
- Связана ли работа со **здоровьем**: режимом сна, занятиями спортом, курением, алкоголем?
- Важна ли **интенсивность**: есть ли порог часов, после которого начинаются проблемы?
- Важен ли **тип** работы: помогает ли работа по специальности и вредит ли работа «не по профилю»?
- Как сами работающие студенты оценивают совмещение: помогает ли работа усваивать материал, страдает ли успеваемость, трудно ли совмещать?
- Что меняется у **одного и того же студента** между 3-м и 4-м курсом, если он начинает или перестаёт работать?

## Используемые данные

**Источник:** лонгитюдное исследование «Модели образовательного поведения студентов в их связи с показателями успешности» (Институт образования НИУ ВШЭ, https://ioe.hse.ru/cshe/project/). Исследование идёт с осени 2022 года в **девяти селективных российских университетах**.

**Как собирались данные.** Выборка многоступенчатая: в каждом университете отбираются укрупнённые направления подготовки, внутри них случайно выбираются учебные группы, и в начале каждого учебного года в этих группах опрашиваются все студенты (тестирование и анкета в компьютерных классах).

**Что есть в массиве:**
- **3-я волна** — 2024/2025 учебный год, студенты 3-го курса;
- **4-я волна** — 2025/2026 учебный год, те же студенты на 4-м курсе;
- не вся анкета, а только переменные о **занятости**, **опыте обучения** и **здоровье**.

**Особенности данных, которые нужно учитывать:**
- **Лонгитюд.** Одни и те же студенты опрошены дважды. Можно сравнивать не только разных студентов, но и изменения у одного человека во времени.
- **Осыпание выборки.** Часть студентов не приходит на следующую волну, и выбывают они не случайно (например, отчислившиеся). Это может искажать результаты.

- **Вопросы с фильтрами.** Часть вопросов задавали не всем: например, часы работы — только работающим, причины мыслей об отчислении — только тем, кто об этом задумывался. Поэтому в данных много закономерных пропусков.
- **Ответы закодированы числами.** Например, `sex = 1` — мужской пол, `sex = 2` — женский. Расшифровки кодов приведены ниже.

## Какие методы я использую и почему

**Словарь переменных**

Названия колонок короткие и непонятные (`ses1`, `edu3_3w`, `pe_3w_2`). Собираю из метаданных таблицу-словарь: формулировка вопроса, волна, тематический блок, заполненность и коды ответов. Ею буду пользоваться весь проект.

```python
def get_wave(column):
    """Определяет волну вопроса по названию колонки."""
    if '_3w' in column or column == 'wave3':
        return '3 волна'
    if '_4w' in column or column == 'wave4':
        return '4 волна'
    return 'без суффикса'


def get_block(column):
    """Определяет тематический блок вопроса по началу названия колонки."""
    column = column.lower()
    if column.startswith('work') or column.startswith('intern') or column.startswith('hours'):
        return 'занятость'
    if column.startswith('mark') or column.startswith('satisfaction') or column.startswith('edu3') or column.startswith('dropout'):
        return 'учёба'
    if column.startswith('sleep') or column.startswith('pe') or column.startswith('smoking') or column.startswith('alco'):
        return 'здоровье'
    return 'общие сведения'


def format_value_labels(labels):
    """Склеивает расшифровку кодов в одну строку, например: '1=Мужской; 2=Женский'."""
    parts = []
    for code, label in labels.items():
        if isinstance(code, float):
            code = int(code)  # 1.0 -> 1
        parts.append(f'{code}={label}')
    return '; '.join(parts)


def build_variable_dictionary(df, meta):
    """Собирает таблицу-словарь: по одной строке на каждую колонку df."""
    rows = []
    for column in df.columns:
        # пустая строка в текстовых колонках — это тоже пропуск
        filled = df[column].replace('', np.nan).notna().sum()
        rows.append({
            'column': column,
            'question': meta.column_names_to_labels.get(column),  # формулировка вопроса
            'wave': get_wave(column),
            'block': get_block(column),
            'n_filled': filled,  # сколько студентов ответили
            'value_labels': format_value_labels(meta.variable_value_labels.get(column, {})),  # расшифровка кодов
        })
    return pd.DataFrame(rows)


variables = build_variable_dictionary(df, meta)
variables.head(5)
```

Словарь сохранён в `variables` — по нему можно найти формулировку любого вопроса. Посмотрю, сколько переменных в каждом тематическом блоке и волне:

## Названия столбцов привести в соответствии с PEP8

Что не так с названиями:
- `Work_major_4w` — заглавная буква;
- вопросы о работе и стажировках **3-й волны идут без суффикса** (`work1`, `work_hours`, `intern1` …), а 4-й — с суффиксом `_4w`. Так легко перепутать волны, поэтому добавляю `_3w`;
- лишние уровни в номерах: `dropout2_3w_1_1` → `dropout2_3w_1`, `pe_3w_1` / `pe_3w_2` → `pe_3w` (занимается ли спортом) и `pe_freq_3w` (как часто);
- `work2_1`, `work2_2`, `work2_3` — это один и тот же вопрос «в какой области вы работаете», который задавали трём группам: работающим полный день, неполный день и с нерегулярными приработками. Переименовываю по группам, а на следующем шаге склею в один столбец;

Чтобы не ошибиться в выводах, называю их `work5_4w_*`. В 3-й волне похожие утверждения (`intern5_*`) действительно относились к стажировке.

### Опечатки и недопустимые коды

Для каждой кодированной колонки сверяю значения со списком вариантов ответа из анкеты (метки значений `.sav`). Любой код вне списка — опечатка.

```python
def find_unknown_codes(df, value_labels):
    """Ищет коды, которых нет среди вариантов ответа в анкете. Возвращает словарь колонка -> лишние коды."""
    unknown = {}
    for column, labels in value_labels.items():
        if column not in df.columns:  # колонку могли удалить при предобработке
            continue
        allowed_codes = set(float(code) for code in labels.keys())
        codes_in_data = set(df[column].dropna().astype(float).unique())
        extra_codes = codes_in_data - allowed_codes
        if extra_codes:
            unknown[column] = sorted(extra_codes)
    return unknown


unknown_codes = find_unknown_codes(df, value_labels)
print('Недопустимые коды:', unknown_codes if unknown_codes else 'не найдены')
```

Все коды соответствуют анкете — опечаток в закрытых вопросах нет. Открытые ответы «Другое (укажите)» (`work2_4w_88`, `work4_4w_88`) — свободный текст, в количественном анализе я их не использую и не правлю.

### Незавершённые анкеты 4-й волны

В описании данных я заметил, что около 600 участников 4-й волны ответили на первые вопросы анкеты (повторный выбор вуза, удовлетворённость, мысли об отчислении, оценки), но не дошли до вопросов о работе. Проверяю, похожи ли их ответы на ответы студентов, которые прошли анкету до конца.

Завершённой считаю анкету, в которой есть ответ на вопрос о работе (`work1_4w`) — с него начинается второй блок анкеты.

```python
participants_4w = df['wave4'] == 1  # участники 4-й волны
incomplete_4w = participants_4w & df['work1_4w'].isna()  # участники, не дошедшие до вопроса о работе

print('Завершённых анкет:', (participants_4w & ~incomplete_4w).sum())
print('Брошенных до вопросов о работе:', incomplete_4w.sum())

# сравниваю доли некоторых ответов в брошенных и завершённых анкетах
checks = {
    'Поступил(а) бы в другой вуз на другую специальность': ('edu3_4w', 4),
    'Только отличные оценки': ('mark_4w', 4),
    'Затрудняюсь оценить удовлетворённость': ('satisfaction_4w', 5),
    'Думал(а) об отчислении': ('dropout1_4w', 2),
}

rows = []
for answer, (column, code) in checks.items():
    incomplete_answers = df.loc[incomplete_4w, column].dropna()
    complete_answers = df.loc[participants_4w & ~incomplete_4w, column].dropna()
    rows.append({
        'ответ': answer,
        'брошенные анкеты, %': round((incomplete_answers == code).mean() * 100, 1),
        'завершённые анкеты, %': round((complete_answers == code).mean() * 100, 1),
    })

pd.DataFrame(rows).set_index('ответ')
```

Для сравнения проверяю 3-ю волну: там до вопросов о работе не дошли только 2% участников, и проблемы нет.

Решение: в брошенных анкетах 4-й волны **заменяю на пропуски все ответы первого блока** (повторный выбор вуза, удовлетворённость, мысли об отчислении и их причины, оценки). Строки не удаляю: у многих из этих студентов есть нормальные ответы 3-й волны.

## Служебные коды ответов

В порядковых шкалах (оценки, удовлетворённость, согласие с утверждениями) есть варианты, которые **не являются точкой шкалы**:
- `88` — «Другое» в вопросе об оценках;
- `5` — «Затрудняюсь ответить» в удовлетворённости и соответствии работы специальности;
- `99` — «Затрудняюсь ответить» в утверждениях о стажировке и работе;
- `3` — «Предпочитаю не отвечать» о курении и алкоголе;
- `7` в материальном положении семьи — «Не было родителей, жил(а) в интернате» (особая ситуация, не ступень шкалы).

Если оставить их как числа, «затрудняюсь» (5) окажется «выше», чем «полностью удовлетворён» (4), и средние будут искажены. Заменяю их на пропуски.

В номинальных вопросах (сфера работы, мотив работы, цель стажировки) `88 = «Другое»` — нормальная категория, её оставляю.

```python
# колонка -> служебные коды, которые не являются точкой шкалы
service_codes = {
    'mark_3w': [88], 'mark_4w': [88],
    'satisfaction_3w': [5], 'satisfaction_4w': [5],
    'work_major_4w': [5],
    'intern_major_4w': [99],
    'ses1': [7, 99],
    'smoking_3w': [3], 'smoking_4w': [3],
    'alco_3w': [3], 'alco_4w': [3],
    'intern5_3w_1': [99], 'intern5_3w_2': [99], 'intern5_3w_3': [99], 'intern5_3w_4': [99],
    'work5_4w_1': [99], 'work5_4w_2': [99], 'work5_4w_3': [99],
}


def replace_service_codes(df, service_codes):
    """Заменяет служебные коды на пропуски. Возвращает таблицу: сколько ответов заменено в каждой колонке."""
    rows = []
    for column, codes in service_codes.items():
        is_service = df[column].isin(codes)
        answered = df[column].notna().sum()
        rows.append({
            'колонка': column,
            'заменено': is_service.sum(),
            'доля от ответивших, %': round(is_service.sum() / answered * 100, 1),
        })
        df.loc[is_service, column] = pd.NA
    return pd.DataFrame(rows).set_index('колонка').sort_values('заменено', ascending=False)


replaced_codes = replace_service_codes(df, service_codes)
replaced_codes.head(5)
```

- **Удовлетворённость обучением:** в 4-й волне «Затрудняюсь ответить» выбрали 9.1% студентов (после очистки незавершённых анкет), а в 3-й волне этого ответа нет совсем. Похоже, в 3-й волне этот вариант не показывали. Разница небольшая, но при сравнении волн по удовлетворённости это нужно помнить.

## Проверить данные на выбросы и адекватность данных. Предположить причину возникновения выбросов и обработать такие значения

Кодированные колонки я уже проверил на допустимые значения — выбросов там быть не может. Доли времени `hours_3w_*` лежат в диапазоне 0–100 и в сумме дают 100%.

Остаются **часы работы на прошлой неделе** (`work_hours_3w`, `work_hours_4w`) — вопрос с открытым числовым ответом. Посмотрю на их статистики.

```python
work_hours_columns = ['work_hours_3w', 'work_hours_4w']
df[work_hours_columns].describe().T
```

Видно три типа проблем:
1. **Отрицательные значения** (−40, −15, −1). Вероятная причина — случайно набранный минус: модуль числа выглядит правдоподобно (40 часов — полная ставка). Беру модуль.
2. **Физически невозможные значения** — больше 168 часов (столько часов в неделе). Это 500, 3383, 10^94 и т.п. — шутливые или случайные ответы. Можно было бы предположить, что человек указал часы **в месяц**, но тогда пришлось бы выбирать границу «неделя/месяц» наугад, а таких ответов единицы — восстанавливать их ненадёжно. Заменяю на пропуск.
3. **Правдоподобные, но очень большие значения** (85–168 часов). Их проверяю по межквартильному размаху.

**Почему 3·IQR, а не 1.5·IQR.** Распределение часов сильно скошено вправо: большинство работает 10–25 часов, но есть студенты с полной ставкой. Граница 1.5·IQR для 3-й волны — около 50 часов, то есть отрезала бы реальных людей, работающих полный день с переработками. Граница 3·IQR отсекает только экстремальные значения, которые не совмещаются с очной учёбой (больше ~12 часов работы каждый день без выходных).

Важно: я **не удаляю студентов**, а заменяю на пропуск только неверное значение часов. Все остальные ответы студента остаются в анализе.

```python
def clean_work_hours(hours, iqr_factor=3, hours_in_week=168):
    """
    Чистит часы работы за неделю:
    1. отрицательные значения -> модуль (случайный минус);
    2. больше 168 часов -> NaN (столько часов в неделе);
    3. выше Q3 + 3*IQR -> NaN (нереалистично при очной учёбе).
    Возвращает очищенные часы и отчёт о том, сколько значений изменено.
    """
    cleaned = hours.copy()

    # 1. отрицательные значения
    negative = cleaned < 0
    cleaned[negative] = cleaned[negative].abs()

    # 2. физически невозможные значения
    impossible = cleaned > hours_in_week
    cleaned[impossible] = np.nan

    # 3. экстремальные значения по межквартильному размаху
    q1 = cleaned.quantile(0.25)
    q3 = cleaned.quantile(0.75)
    upper_bound = q3 + iqr_factor * (q3 - q1)
    extreme = cleaned > upper_bound
    cleaned[extreme] = np.nan

    changed = negative | impossible | extreme
    report = {
        'ответов': hours.notna().sum(),
        'отрицательных -> модуль': negative.sum(),
        '> 168 ч -> NaN': impossible.sum(),
        'граница 3*IQR, ч': round(upper_bound, 1),
        '> 3*IQR -> NaN': extreme.sum(),
        'доля изменённых, %': round(changed.sum() / hours.notna().sum() * 100, 1),
    }
    return cleaned, report


df['work_hours_3w'], report_3w = clean_work_hours(df['work_hours_3w'])
df['work_hours_4w'], report_4w = clean_work_hours(df['work_hours_4w'])

pd.DataFrame({'work_hours_3w': report_3w, 'work_hours_4w': report_4w})
```

## Подготовка данных для анализа: одна строка — один студент в одной волне

Сейчас ответы 3-й и 4-й волн лежат в разных колонках одной строки. Чтобы сравнивать волны на одном графике и не писать каждый расчёт дважды, я собираю из `df` таблицу `df_long` в «длинном» формате: **одна строка — один студент в одной волне**.
- В неё попадают вопросы, которые задавались в обеих волнах, и общие сведения о студенте.
- Суффикс волны из названия убирается, а сама волна записывается в колонку `wave` (3 или 4).
- `student_id` — номер строки студента в `df`. По нему можно найти одного и того же студента в обеих волнах.

Вопросы, которые есть только в одной волне (распределение времени, оценки совмещения работы и учёбы и т.п.), я анализирую по `df`.

```python
# общие сведения о студенте — одинаковые в обеих волнах
general_columns = ['bigmajor', 'funding', 'sex', 'ses1']

# вопросы, которые задавались в обеих волнах (без суффикса волны)
common_questions = ['edu3', 'satisfaction', 'dropout1', 'mark',
                    'work1', 'work_intentions', 'work_hours', 'work2', 'work4',
                    'intern1', 'intern2', 'intern3',
                    'sleep1', 'sleep2', 'sleep3', 'sleep4', 'pe', 'smoking', 'alco',
                    'dropout2_1', 'dropout2_2', 'dropout2_3', 'dropout2_4', 'dropout2_5', 'dropout2_6', 'dropout2_7']


def wave_column(question, wave):
    """Название колонки вопроса в нужной волне: mark -> mark_3w, dropout2_1 -> dropout2_3w_1."""
    if question.startswith('dropout2_'):
        number = question.split('_')[1]
        return f'dropout2_{wave}_{number}'
    return f'{question}_{wave}'


def to_long(df):
    """Собирает длинную таблицу: участники 3-й волны, затем участники 4-й волны, вопросы без суффикса."""
    parts = []
    for wave in [3, 4]:
        # колонки этой волны -> названия без суффикса
        renames = {}
        for question in common_questions:
            renames[wave_column(question, f'{wave}w')] = question

        participants = df[df[f'wave{wave}'] == 1]  # только участники волны
        part = participants[general_columns + list(renames.keys())].copy()
        part = part.rename(columns=renames)
        part['student_id'] = participants.index  # номер строки студента в df
        part['wave'] = wave
        parts.append(part)

    df_long = pd.concat(parts, ignore_index=True)
    # переставляю student_id и wave в начало таблицы, так удобнее читать
    first_columns = ['student_id', 'wave']
    other_columns = [c for c in df_long.columns if c not in first_columns]
    return df_long[first_columns + other_columns]


df_long = to_long(df)
print(f'Строк в df_long: {len(df_long)}')
df_long.head(5)
```

Что получили:
- `df_long` — новая таблица для анализа: одна строка = один студент в одной волне. Колонка `wave` — номер волны (3 или 4), `student_id` — номер строки этого студента в `df`;
- `df` при этом не меняется: в нём по-прежнему одна строка на студента и все вопросы, включая те, что были только в одной волне;
- `general_columns` и `common_questions` — списки колонок, которые попали в `df_long`.

- `share_table` — распределение ответов в процентах внутри каждой группы (например, внутри каждой волны);
- `plot_shares` — столбчатая диаграмма по такой таблице.

```python
# короткие подписи кодов для графиков: колонка -> {код: подпись}
SHORT_LABELS = {
    'bigmajor': {1: 'Мат. и естеств. науки', 2: 'Инженерные и техн.', 3: 'Медицина', 4: 'Сельское хоз-во',
                 5: 'Науки об обществе', 6: 'Педагогика', 7: 'Гуманитарные', 8: 'Искусство'},
    'funding': {1: 'Бюджет', 2: 'Платное', 3: 'Целевое'},
    'sex': {1: 'Мужской', 2: 'Женский'},
    'ses1': {1: 'Не хватало на еду', 2: 'Хватало только на еду', 3: 'Трудно с одеждой',
             4: 'Трудно с техникой', 5: 'Трудно с автомобилем', 6: 'Обеспечены'},
    'work1': {1: 'Полный день', 2: 'Неполный день', 3: 'Приработки', 4: 'Не работает'},
    'work2': {1: 'Образование', 2: 'Наука', 3: 'Бизнес (наёмный)', 4: 'Бизнес (свой)', 5: 'Госуправление',
              6: 'Обществ. организации', 7: 'СМИ', 8: 'Фриланс', 88: 'Другое'},
    'work4': {1: 'Оплатить обучение', 2: 'Содержать себя', 3: 'Доп. доход', 4: 'Опыт работы',
              5: 'Интерес к сфере', 88: 'Другое'},
    'mark': {1: 'В основном 3', 2: 'В основном 4', 3: '4 и 5', 4: 'Только 5'},
    'satisfaction': {1: 'Совсем не удовлетворён', 2: 'Скорее не удовлетворён',
                     3: 'Скорее удовлетворён', 4: 'Полностью удовлетворён'},
    'dropout1': {1: 'Не думал', 2: 'Думал, но остаётся', 3: 'Намерен отчислиться'},
    'edu3': {1: 'Тот же вуз и специальность', 2: 'Тот же вуз, другая спец.', 3: 'Другой вуз, та же спец.',
             4: 'Другой вуз и спец.', 5: 'Не стал бы поступать'},
    'agree': {1: 'Абсолютно не согласен', 2: 'Скорее не согласен', 3: 'Скорее согласен', 4: 'Полностью согласен'},
    'major_match': {1: 'Совсем не соотв.', 2: 'Скорее не соотв.', 3: 'Скорее соотв.', 4: 'Полностью соотв.'},
    'yes_no': {1: 'Да', 2: 'Нет'},
}
WAVE_LABELS = {3: '3 волна (3 курс)', 4: '4 волна (4 курс)'}
WORK_ORDER = ['Полный день', 'Неполный день', 'Приработки', 'Не работает']  # порядок групп на графиках


def share_table(data, column, by='wave', labels=None):
    """
    Считает распределение ответов в процентах внутри каждой группы (по умолчанию — внутри каждой волны).
    Коды ответов заменяются короткими подписями из SHORT_LABELS.
    """
    if labels is None:
        labels = SHORT_LABELS[column]
    answers = data[column].map(labels)  # коды -> подписи
    table = pd.crosstab(answers, data[by], normalize='columns') * 100  # normalize='columns': проценты внутри группы

    # порядок строк как в анкете (а не по алфавиту)
    order = []
    for label in labels.values():
        if label in table.index:
            order.append(label)
    return table.loc[order].round(1)


def plot_shares(table, title, ax=None):
    """Рисует горизонтальную столбчатую диаграмму по таблице из share_table."""
    if ax is None:
        fig, ax = plt.subplots(figsize=(10, 0.6 * len(table) + 1.5))
    table = table.rename(columns=WAVE_LABELS)  # 3 -> «3 волна (3 курс)»
    table.plot.barh(ax=ax, width=0.75)
    ax.invert_yaxis()  # первый вариант ответа — сверху
    ax.set_title(title)
    ax.set_xlabel('% студентов')
    ax.set_ylabel('')
    ax.legend(loc='lower right', fontsize=9)
    return ax
```

### Причины мыслей об отчислении

### Режим сна

Время отхода ко сну и подъёма записано кодами (например, отход ко сну: 1 — 20:00, 5 — 00:00). Чтобы считать продолжительность сна, перевожу коды в часы:
- отход ко сну: код 1 = 20:00, шаг 1 час. Время после полуночи считаю как 24, 25, … часов (00:00 = 24, 2:00 = 26), чтобы ночь не «разрывалась»;
- подъём: код 1 = 3:00, шаг 1 час. Подъём всегда на следующий день, поэтому прибавляю 24 часа;
- продолжительность сна = подъём − отход ко сну.

Продолжительность меньше 3 или больше 16 часов считаю невозможной для обычного режима (например, «лёг в 6:00, встал в 3:00») и заменяю на пропуск.

```python
def bedtime_hours(code):
    """Код времени отхода ко сну -> час на непрерывной шкале: код 1 = 20:00 -> 20, код 5 = 00:00 -> 24, код 7 = 2:00 -> 26."""
    return code + 19


def wake_hours(code):
    """Код времени подъёма -> час следующего дня на той же шкале: код 1 = 3:00 -> 27, код 5 = 7:00 -> 31."""
    return code + 2 + 24


def sleep_duration(bed_code, wake_code, min_hours=3, max_hours=16):
    """Продолжительность сна в часах; значения вне диапазона 3–16 часов заменяются на NaN."""
    duration = wake_hours(wake_code) - bedtime_hours(bed_code)
    duration = duration.astype('float')
    duration[(duration < min_hours) | (duration > max_hours)] = np.nan
    return duration


# новые колонки в df_long
df_long['bedtime_weekday'] = bedtime_hours(df_long['sleep1']).astype('float')  # во сколько ложится в будни
df_long['sleep_weekday'] = sleep_duration(df_long['sleep1'], df_long['sleep3'])  # сколько спит в будни
df_long['sleep_weekend'] = sleep_duration(df_long['sleep2'], df_long['sleep4'])  # сколько спит в выходные

answered_sleep = df_long['sleep1'].notna() & df_long['sleep3'].notna()
impossible_sleep = answered_sleep & df_long['sleep_weekday'].isna()
print(f'Ответов о сне в будни: {answered_sleep.sum()}, невозможная продолжительность -> NaN: {impossible_sleep.sum()}')

df_long.groupby('wave')[['sleep_weekday', 'sleep_weekend']].describe().T.round(1).head(10)
```

Что получили: в `df_long` появились три новые колонки:
- `bedtime_weekday` — во сколько студент ложится спать в будни (24 = полночь, 26 = 2:00 ночи);
- `sleep_weekday` — сколько часов он спит в будни;
- `sleep_weekend` — сколько часов он спит в выходные.

Это описательное сравнение без проверки значимости: оно показывает, где искать различия. Проверю их в разделе с гипотезами.

```python
def share_of(answers, codes):
    """Доля (в %) ответов с нужными кодами среди всех, кто ответил на вопрос."""
    answered = answers.dropna()
    return round(answered.isin(codes).mean() * 100, 1)


df_long['work_status'] = df_long['work1'].map(SHORT_LABELS['work1'])

rows = []
for wave in [3, 4]:
    for status in WORK_ORDER:
        group = df_long[(df_long['wave'] == wave) & (df_long['work_status'] == status)]
        rows.append({
            'wave': wave,
            'work_status': status,
            'Студентов': len(group),
            'Хорошие оценки, %': share_of(group['mark'], [3, 4]),
            'Удовлетворены обучением, %': share_of(group['satisfaction'], [3, 4]),
            'Думали об отчислении, %': share_of(group['dropout1'], [2, 3]),
            'Выбрали бы тот же вуз и спец., %': share_of(group['edu3'], [1]),
            'Ложатся в 2:00 и позже, %': share_of(group['sleep1'], [7, 8, 9, 10, 11]),  # коды 7–11 = 2:00–6:00
            'Занимаются спортом, %': share_of(group['pe'], [1]),
            'Сон в будни, ч': round(group['sleep_weekday'].mean(), 2),
        })

work_vs_outcomes = pd.DataFrame(rows)
work_vs_outcomes
```

# Проведение расчётов и исследований

## Новые признаки

Все новые признаки добавляю в `df_long`:
- `works` — работает ли студент: 1 — да (любой тип занятости), 0 — нет;
- `work_level` — интенсивность занятости по порядку: 0 — не работает, 1 — приработки, 2 — неполный день, 3 — полный день. В анкете коды идут наоборот (1 — полный день, 4 — не работает), поэтому перекодирую, чтобы «больше» значило «больше работы»;
- `hours_group` — группа по часам работы за прошлую неделю: 0, 1–10, 11–20, 21–30, больше 30 часов;
- признаки-флаги для результатов (1 — да, 0 — нет): `good_marks` (оценки «4 и 5» или «только 5»), `satisfied` (скорее или полностью удовлетворён обучением), `dropout_thoughts` (думал об отчислении), `short_sleep` (спит меньше 7 часов в будни), `late_bedtime` (ложится в 2:00 и позже), `does_sport` (занимается спортом).

```python
# 1. работает ли студент
df_long['works'] = np.nan
df_long.loc[df_long['work1'].isin([1, 2, 3]), 'works'] = 1
df_long.loc[df_long['work1'] == 4, 'works'] = 0

# 2. интенсивность занятости: перекодирую так, чтобы больше = больше работы
work_level_codes = {4: 0, 3: 1, 2: 2, 1: 3}  # код в анкете -> уровень
df_long['work_level'] = df_long['work1'].map(work_level_codes).astype('float')


# 3. группа по часам работы
def hours_group(hours):
    """Группа по числу часов работы за прошлую неделю."""
    if pd.isna(hours):
        return np.nan
    if hours == 0:
        return '0 ч'
    if hours <= 10:
        return '1–10 ч'
    if hours <= 20:
        return '11–20 ч'
    if hours <= 30:
        return '21–30 ч'
    return '31+ ч'


HOURS_ORDER = ['0 ч', '1–10 ч', '11–20 ч', '21–30 ч', '31+ ч']  # порядок групп на графиках
df_long['hours_group'] = df_long['work_hours'].apply(hours_group)


# 4. флаги результатов
def make_flag(answers, codes):
    """1 — ответ входит в codes, 0 — не входит, NaN — студент не ответил."""
    flag = answers.isin(codes).astype('float')
    flag[answers.isna()] = np.nan
    return flag


df_long['good_marks'] = make_flag(df_long['mark'], [3, 4])
df_long['satisfied'] = make_flag(df_long['satisfaction'], [3, 4])
df_long['dropout_thoughts'] = make_flag(df_long['dropout1'], [2, 3])
df_long['late_bedtime'] = make_flag(df_long['sleep1'], [7, 8, 9, 10, 11])  # коды 7–11 = 2:00–6:00
df_long['does_sport'] = make_flag(df_long['pe'], [1])

df_long['short_sleep'] = (df_long['sleep_weekday'] < 7).astype('float')
df_long.loc[df_long['sleep_weekday'].isna(), 'short_sleep'] = np.nan

new_features = ['works', 'work_level', 'hours_group', 'good_marks', 'satisfied',
                'dropout_thoughts', 'short_sleep', 'late_bedtime', 'does_sport']
df_long[['student_id', 'wave', 'work1', 'work_hours'] + new_features].head(5)
```

Что получили: в `df_long` 9 новых колонок (список — в переменной `new_features`). Флаги удобны тем, что среднее по ним — это сразу доля студентов: например, среднее `good_marks` по группе = доля студентов с хорошими оценками в этой группе.

Проверяю, сколько студентов попало в каждую группу по часам работы:

```python
pd.crosstab(df_long['hours_group'], df_long['wave']).loc[HOURS_ORDER]
```

**Вывод:**
- В группе «0 ч» в основном неработающие студенты. В 4-й волне туда же попал 51 студент, который работает, но на прошлой неделе не работал (почти все — с нерегулярными приработками). Поэтому в 4-й волне в этой группе 524 студента, а неработающих — 473.
- Самые малочисленные группы — 21–30 и больше 30 часов (56–74 студента). Различия в них нужно проверять статистически, а не делать выводы только по графикам.

## Интенсивность занятости: есть ли порог по часам работы

Для каждой группы по часам считаю долю студентов с хорошими оценками, с мыслями об отчислении, с коротким сном и среднюю продолжительность сна. Если работа начинает мешать после какого-то количества часов, на графиках будет видно «перелом».

```python
outcome_flags = {
    'good_marks': 'Хорошие оценки, %',
    'dropout_thoughts': 'Думали об отчислении, %',
    'satisfied': 'Удовлетворены обучением, %',
    'short_sleep': 'Спят меньше 7 часов в будни, %',
}

rows = []
for wave in [3, 4]:
    for group_name in HOURS_ORDER:
        group = df_long[(df_long['wave'] == wave) & (df_long['hours_group'] == group_name)]
        row = {'wave': wave, 'hours_group': group_name, 'Студентов': len(group)}
        for flag, label in outcome_flags.items():
            row[label] = round(group[flag].mean() * 100, 1)  # среднее флага = доля
        row['Сон в будни, ч'] = round(group['sleep_weekday'].mean(), 2)
        rows.append(row)

hours_vs_outcomes = pd.DataFrame(rows)
hours_vs_outcomes
```

## Тип работы: по специальности или нет, ради денег или ради опыта (4 волна)

Вопрос о соответствии работы специальности задавали только в 4-й волне, поэтому этот расчёт делаю по работающим студентам 4 курса (`workers_4w` из раздела анализа).

Делю работающих на группы двумя способами:
- **по соответствию специальности:** «по специальности» (скорее или полностью соответствует) и «не по специальности» (скорее или совсем не соответствует);
- **по мотиву:** «ради денег» (оплатить обучение, содержать себя, дополнительный доход) и «ради развития» (опыт работы, интерес к сфере).

Для каждой группы считаю, как студенты оценивают совмещение, и сравниваю их оценки, мысли об отчислении и сон.

```python
workers_4w = df[df['work1_4w'].isin([1, 2, 3])].copy()

# соответствие специальности: коды 3–4 — по специальности, 1–2 — нет
workers_4w['major_match'] = np.nan
workers_4w.loc[workers_4w['work_major_4w'].isin([3, 4]), 'major_match'] = 'По специальности'
workers_4w.loc[workers_4w['work_major_4w'].isin([1, 2]), 'major_match'] = 'Не по специальности'

# мотив работы: коды 1–3 — деньги, 4–5 — развитие («другое» не использую)
workers_4w['motive'] = np.nan
workers_4w.loc[workers_4w['work4_4w'].isin([1, 2, 3]), 'motive'] = 'Ради денег'
workers_4w.loc[workers_4w['work4_4w'].isin([4, 5]), 'motive'] = 'Ради развития'

# сон 4-й волны — теми же функциями, что и раньше
workers_4w['sleep_weekday'] = sleep_duration(workers_4w['sleep1_4w'], workers_4w['sleep3_4w'])


def describe_work_group(group):
    """Основные показатели для группы работающих студентов 4 курса."""
    return {
        'Студентов': len(group),
        'Часов работы (медиана)': group['work_hours_4w'].median(),
        'Работа помогает учиться, %': share_of(group['work5_4w_1'], [3, 4]),
        'Страдает успеваемость, %': share_of(group['work5_4w_2'], [3, 4]),
        'Трудно совмещать, %': share_of(group['work5_4w_3'], [3, 4]),
        'Хорошие оценки, %': share_of(group['mark_4w'], [3, 4]),
        'Думали об отчислении, %': share_of(group['dropout1_4w'], [2, 3]),
        'Сон в будни, ч': round(group['sleep_weekday'].mean(), 2),
    }


rows = []
for column in ['major_match', 'motive']:
    for group_name, group in workers_4w.groupby(column):
        row = {'группа': group_name}
        row.update(describe_work_group(group))
        rows.append(row)

work_type_table = pd.DataFrame(rows).set_index('группа')
work_type_table
```

Что получили:

- `work_type_table` — сравнение этих групп по оценкам совмещения, учёбе и сну.

`share_of` — функция из раздела анализа: доля нужных ответов среди ответивших.

```python
statements = ['Работа помогает учиться, %', 'Страдает успеваемость, %', 'Трудно совмещать, %']

fig, axes = plt.subplots(1, 2, figsize=(16, 5), sharey=True)
work_type_table.loc[['По специальности', 'Не по специальности'], statements].T.plot.bar(ax=axes[0], rot=0)
axes[0].set_title('Соответствие работы специальности')
work_type_table.loc[['Ради развития', 'Ради денег'], statements].T.plot.bar(ax=axes[1], rot=0)
axes[1].set_title('Мотив работы')
for ax in axes:
    ax.set_ylabel('% согласных')
    ax.legend(title='')
    for container in ax.containers:
        ax.bar_label(container, fmt='%.0f', padding=2)
plt.tight_layout()
plt.show()
```

- То же с мотивом: те, кто работает **ради опыта и интереса**, чаще видят пользу для учёбы (71% против 53%), реже жалуются на падение успеваемости (20% против 31%) и трудности совмещения (25% против 37%), лучше учатся (66% против 55% хороших оценок) и немного больше спят.

## Лонгитюд: что меняется у одних и тех же студентов

Студенты, которые есть в обеих волнах, позволяют посмотреть на изменения у одного и того же человека. По тому, работал ли студент на 3 и на 4 курсе, делю их на четыре траектории:
- **не работал → не работает**;
- **начал работать** (не работал на 3 курсе, работает на 4-м);
- **перестал работать** (работал на 3 курсе, не работает на 4-м);
- **работал → работает**.

Если работа мешает учёбе и сну, у тех, кто начал работать, показатели должны ухудшиться сильнее, чем у тех, кто не работал оба года.

Беру только студентов, которые ответили на вопрос о работе в обеих волнах.

**Почему так собираю таблицу.** Чтобы сравнить студента с самим собой, его ответы 3-й и 4-й волн должны оказаться в одной строке. Беру из `df_long` строки каждой волны и соединяю их по `student_id` методом `merge` — это как ВПР в Excel: строки с одинаковым `student_id` склеиваются.

```python
panel_columns = ['student_id', 'works', 'good_marks', 'dropout_thoughts', 'sleep_weekday']

wave3 = df_long.loc[df_long['wave'] == 3, panel_columns]
wave4 = df_long.loc[df_long['wave'] == 4, panel_columns]

# соединяю ответы одного студента из двух волн в одну строку
panel = wave3.merge(wave4, on='student_id', suffixes=('_3w', '_4w'))
panel = panel.dropna(subset=['works_3w', 'works_4w'])  # ответил о работе в обеих волнах
print(f'Студентов в обеих волнах с ответом о работе: {len(panel)}')


def trajectory(row):
    """Траектория занятости студента между 3 и 4 курсом."""
    if row['works_3w'] == 0 and row['works_4w'] == 0:
        return 'Не работал -> не работает'
    if row['works_3w'] == 0 and row['works_4w'] == 1:
        return 'Начал работать'
    if row['works_3w'] == 1 and row['works_4w'] == 0:
        return 'Перестал работать'
    return 'Работал -> работает'


TRAJECTORY_ORDER = ['Не работал -> не работает', 'Начал работать', 'Перестал работать', 'Работал -> работает']
panel['trajectory'] = panel.apply(trajectory, axis=1)
panel['trajectory'].value_counts().loc[TRAJECTORY_ORDER]
```

Что получили: `panel` — таблица студентов из обеих волн, одна строка на студента. Колонки с суффиксом `_3w` — показатели на 3 курсе, с `_4w` — на 4 курсе, `trajectory` — траектория занятости.

Считаю для каждой траектории показатели на 3 и на 4 курсе и их изменение:

```python
rows = []
for name in TRAJECTORY_ORDER:
    group = panel[panel['trajectory'] == name]
    rows.append({
        'траектория': name,
        'Студентов': len(group),
        'Хорошие оценки 3к, %': round(group['good_marks_3w'].mean() * 100, 1),
        'Хорошие оценки 4к, %': round(group['good_marks_4w'].mean() * 100, 1),
        'Думали об отчислении 3к, %': round(group['dropout_thoughts_3w'].mean() * 100, 1),
        'Думали об отчислении 4к, %': round(group['dropout_thoughts_4w'].mean() * 100, 1),
        'Сон в будни 3к, ч': round(group['sleep_weekday_3w'].mean(), 2),
        'Сон в будни 4к, ч': round(group['sleep_weekday_4w'].mean(), 2),
    })

trajectory_table = pd.DataFrame(rows).set_index('траектория')

# изменение показателей с 3 на 4 курс
trajectory_table['Изменение: хорошие оценки, п.п.'] = trajectory_table['Хорошие оценки 4к, %'] - trajectory_table['Хорошие оценки 3к, %']
trajectory_table['Изменение: отчисление, п.п.'] = trajectory_table['Думали об отчислении 4к, %'] - trajectory_table['Думали об отчислении 3к, %']
trajectory_table['Изменение: сон, ч'] = trajectory_table['Сон в будни 4к, ч'] - trajectory_table['Сон в будни 3к, ч']
trajectory_table.round(2)
```

```python
changes = ['Изменение: хорошие оценки, п.п.', 'Изменение: отчисление, п.п.', 'Изменение: сон, ч']

fig, axes = plt.subplots(1, 3, figsize=(20, 5))
for ax, column in zip(axes, changes):
    values = trajectory_table[column]
    colors = ['#2a6f97' if value >= 0 else '#c0392b' for value in values]  # рост — синий, падение — красный
    ax.barh(values.index, values, color=colors)
    ax.axvline(0, color='black', linewidth=0.8)
    ax.invert_yaxis()
    ax.set_title(column)
for ax in axes[1:]:
    ax.set_yticklabels([])
plt.tight_layout()
plt.show()
```

**Вывод** (566 студентов, ответивших о работе в обеих волнах):
- **Начало работы не ухудшило показатели.** У тех, кто начал работать на 4 курсе, доля хороших оценок выросла на 7 п.п., почти как у тех, кто не работал оба года (+8 п.п.). Сон вырос одинаково (+0.3 часа), мысли об отчислении не стали чаще.
- **У тех, кто работал оба года, динамика хуже:** доля хороших оценок не выросла (−2 п.п., а у остальных групп +3…+8 п.п.), а мысли об отчислении остались на высоком уровне — около 50% против 35–37% у остальных. Длительное совмещение может накапливать трудности.
- **Те, кто перестал работать,** реже думают об отчислении (−6 п.п.) и изначально учились лучше всех (70% хороших оценок).
- У всех групп сон на 4 курсе немного больше, чем на 3-м. Видимо, на 4 курсе меньше учебной нагрузки.
- Группы небольшие (71–248 студентов), значимость изменений проверю в разделе с гипотезами.

# Корреляционный анализ

Проверяю, как связаны между собой показатели занятости, учёбы и здоровья.

**Какие переменные беру** (все из `df_long`):
- занятость: `work_level` (0–3, интенсивность), `work_hours` (часы в неделю);
- учёба: `mark` (1–4, чем больше, тем лучше оценки), `satisfaction` (1–4), `dropout1` (1–3, чем больше, тем серьёзнее мысли об отчислении);
- здоровье: `sleep_weekday` (часы сна в будни), `bedtime_weekday` (время отхода ко сну), `does_sport`, `smoking`, `alco` (1 — да, 0 — нет).

**Какой метод корреляции выбрать.** Метод Пирсона (он используется в `.corr()` по умолчанию) подходит только для количественных переменных с нормальным распределением. В этих данных:
- большинство переменных — порядковые шкалы (оценки, удовлетворённость, уровень занятости), для них Пирсон некорректен;
- количественные переменные (часы работы, часы сна) нужно проверить на нормальность.

Проверяю нормальность тестом Шапиро–Уилка.

**Почему тест Шапиро–Уилка.** От нормальности распределения зависит выбор метода корреляции и тестов. Тест Шапиро–Уилка проверяет гипотезу «распределение нормальное»: если p-value < 0.05, нормальность отвергается. Это один из самых мощных тестов нормальности для выборок такого размера (до нескольких тысяч наблюдений).

```python
# курение и алкоголь: 1 — да, 2 — нет -> перекодирую в 1 и 0, как спорт
df_long['smokes'] = make_flag(df_long['smoking'], [1])
df_long['drinks'] = make_flag(df_long['alco'], [1])

for column in ['work_hours', 'sleep_weekday', 'bedtime_weekday']:
    values = df_long[column].dropna()
    statistic, p_value = shapiro(values)
    print(f'{column}: W = {statistic:.3f}, p-value = {p_value:.2e} -> {"нормальное" if p_value > 0.05 else "НЕ нормальное"}')
```

Ни одна количественная переменная не распределена нормально (p-value много меньше 0.05). Значит, **использую корреляцию Спирмена**: она работает с рангами, не требует нормальности и подходит для порядковых шкал. Для бинарных переменных (спорт, курение, алкоголь) Спирмен тоже применим — он показывает, чаще ли встречается «1» при больших значениях другой переменной.

Считаю корреляции отдельно для каждой волны, чтобы один и тот же студент не попадал в расчёт дважды.

```python
correlation_columns = {
    'work_level': 'Интенсивность работы',
    'work_hours': 'Часы работы',
    'mark': 'Оценки',
    'satisfaction': 'Удовлетворённость',
    'dropout1': 'Мысли об отчислении',
    'sleep_weekday': 'Сон в будни',
    'bedtime_weekday': 'Время отхода ко сну',
    'does_sport': 'Спорт',
    'smokes': 'Курение',
    'drinks': 'Алкоголь',
}

correlations = {}
for wave in [3, 4]:
    data = df_long.loc[df_long['wave'] == wave, list(correlation_columns.keys())].astype('float')
    correlations[wave] = data.corr(method='spearman').rename(index=correlation_columns, columns=correlation_columns)

fig, axes = plt.subplots(1, 2, figsize=(22, 9))
for ax, wave in zip(axes, [3, 4]):
    mask = np.triu(np.ones_like(correlations[wave], dtype=bool))  # скрываю верхний треугольник: он дублирует нижний
    sns.heatmap(correlations[wave], mask=mask, annot=True, fmt='.2f', cmap='RdBu_r', vmin=-1, vmax=1,
                cbar=False, ax=ax)
    ax.set_title(f'Корреляция Спирмена, {WAVE_LABELS[wave]}')
plt.tight_layout()
plt.show()
```

Что получили: `correlations` — словарь из двух матриц корреляции: `correlations[3]` для 3-й волны и `correlations[4]` для 4-й.

Для интерпретации силы связи использую шкалу Чеддока (по модулю коэффициента): до 0.1 — связи практически нет, 0.1–0.3 — слабая, 0.3–0.5 — умеренная, 0.5–0.7 — заметная, 0.7–0.9 — высокая, больше 0.9 — очень высокая.

Отдельно проверяю значимость связей **занятости** с показателями учёбы и здоровья: коэффициент может быть маленьким, но значимым, если студентов много.

```python
def chaddock(r):
    """Сила связи по шкале Чеддока."""
    r = abs(r)
    if r < 0.1:
        return 'практически нет'
    if r < 0.3:
        return 'слабая'
    if r < 0.5:
        return 'умеренная'
    if r < 0.7:
        return 'заметная'
    if r < 0.9:
        return 'высокая'
    return 'очень высокая'


outcomes_for_test = ['mark', 'satisfaction', 'dropout1', 'sleep_weekday', 'bedtime_weekday', 'does_sport', 'smokes', 'drinks']

rows = []
for wave in [3, 4]:
    data = df_long[df_long['wave'] == wave]
    for work_column in ['work_level', 'work_hours']:
        for outcome in outcomes_for_test:
            pair = data[[work_column, outcome]].dropna().astype('float')  # только студенты, ответившие на оба вопроса
            r, p_value = st.spearmanr(pair[work_column], pair[outcome])
            rows.append({
                'волна': wave,
                'занятость': correlation_columns[work_column],
                'показатель': correlation_columns[outcome],
                'n': len(pair),
                'r Спирмена': round(r, 3),
                'p-value': round(p_value, 4),
                'значима (p < 0.05)': 'да' if p_value < 0.05 else 'нет',
                'сила по Чеддоку': chaddock(r),
            })

work_correlations = pd.DataFrame(rows)
work_correlations[work_correlations['значима (p < 0.05)'] == 'да']
```

Что получили: `work_correlations` — таблица всех пар «занятость — показатель» с коэффициентом Спирмена, p-value и силой связи. Выше выведены только значимые пары.

**Вывод по корреляциям:**
- **Связи занятости с учёбой и здоровьем слабые или практически отсутствуют** (все |r| < 0.2 по шкале Чеддока). Работа сама по себе не определяет ни оценки, ни удовлетворённость.
- **С оценками и удовлетворённостью** ни интенсивность, ни часы работы значимо не связаны ни в одной волне.
- **Значимые, но слабые связи:**
  - чем больше студент работает, тем **меньше спит** в будни (r = −0.10, 3 курс);
  - тем чаще **думает об отчислении** (r = 0.08, 3 курс);
  - тем чаще **курит** (r = 0.15–0.17) и **употребляет алкоголь** (r ≈ 0.1) — возможно, это связано с образом жизни и средой на работе;
  - на 4 курсе работающие чаще **занимаются спортом** (r = 0.13–0.15).
- **Самые сильные связи — не про работу:** удовлетворённость и мысли об отчислении (r = −0.37 и −0.24, умеренная/слабая), время отхода ко сну и продолжительность сна (r = −0.43, умеренная), курение и алкоголь (r = 0.34–0.37, умеренная). Интенсивность и часы работы почти совпадают (r = 0.89–0.96), поэтому в регрессию буду брать только одну из них.
- Корреляция показывает линейную (монотонную) связь. Из расчётов выше видно, что эффект работы нелинейный: проблемы появляются только при большой нагрузке и при работе не по профилю. Поэтому в гипотезах я буду сравнивать группы, а не только смотреть на коэффициенты.

# **Гипотезы**

В анализе и расчётах я увидел различия между группами студентов. Теперь проверяю, **статистически значимы** ли они, или это может быть случайностью.

**Общие правила для всех гипотез:**
- уровень значимости — **α = 0.05**: если p-value < 0.05, отвергаю нулевую гипотезу H0;
- каждую гипотезу проверяю **отдельно в каждой волне**: один и тот же студент не должен попадать в одну проверку дважды;
- в каждую проверку беру только студентов, **ответивших на оба нужных вопроса** (например, и о работе, и об оценках). Остальные строки не участвуют, потому что сравнивать по пропуску нельзя;
- **выбор теста:**
  - порядковые и не нормально распределённые величины (оценки по шкале 1–4, часы сна) сравниваю **U-критерием Манна–Уитни**. Он не требует нормальности (её отсутствие я проверил в корреляционном анализе);
  - доли (например, «думал об отчислении — да/нет» в разных группах) сравниваю **критерием хи-квадрат**. Его условие — ожидаемая частота в каждой клетке таблицы не меньше 5, я проверяю это в каждом тесте. Силу связи оцениваю **V Крамера** (0.1 — слабая, 0.3 — средняя, 0.5 — сильная);
  - изменения у одних и тех же студентов (3 курс → 4 курс) проверяю **критерием Уилкоксона** для связанных выборок.

Чтобы не повторять один и тот же код, пишу две функции для тестов.

```python
def chi2_test(data, group_column, flag_column):
    """
    Критерий хи-квадрат: связана ли группа студента с флагом (1/0).
    Возвращает словарь с результатами, включая минимальную ожидаемую частоту (условие теста — не меньше 5)
    и V Крамера (сила связи).
    """
    pair = data[[group_column, flag_column]].dropna()  # только ответившие на оба вопроса
    table = pd.crosstab(pair[group_column], pair[flag_column])  # таблица сопряжённости
    chi2, p_value, dof, expected = st.chi2_contingency(table)

    n = table.values.sum()
    cramers_v = np.sqrt(chi2 / (n * (min(table.shape) - 1)))
    return {
        'n': n,
        'хи-квадрат': round(chi2, 2),
        'p-value': round(p_value, 4),
        'мин. ожидаемая частота': round(expected.min(), 1),
        'V Крамера': round(cramers_v, 3),
        'H0': 'отвергаю' if p_value < 0.05 else 'не отвергаю',
    }


def mann_whitney_test(values_1, values_2):
    """U-критерий Манна–Уитни для двух независимых групп. Пропуски убираются."""
    values_1 = values_1.dropna().astype('float')
    values_2 = values_2.dropna().astype('float')
    statistic, p_value = st.mannwhitneyu(values_1, values_2, alternative='two-sided')
    return {
        'n1': len(values_1),
        'n2': len(values_2),
        'медиана 1': values_1.median(),
        'медиана 2': values_2.median(),
        'среднее 1': round(values_1.mean(), 2),
        'среднее 2': round(values_2.mean(), 2),
        'U': statistic,
        'p-value': round(p_value, 4),
        'H0': 'отвергаю' if p_value < 0.05 else 'не отвергаю',
    }
```

Что получили: две функции.
- `chi2_test(data, group_column, flag_column)` — хи-квадрат для групп и флага 0/1;
- `mann_whitney_test(values_1, values_2)` — Манн–Уитни для двух групп значений.

Каждая возвращает словарь с результатами теста. Из этих словарей я собираю таблицы.

## Гипотеза 1. Различаются ли оценки работающих и неработающих студентов?

Оценки — порядковая переменная (1 — в основном тройки … 4 — только пятёрки), поэтому использую U-критерий Манна–Уитни.

**H0:** распределение оценок у работающих и неработающих студентов одинаковое.
**H1:** распределение оценок различается.

Подвыборка: участники волны, ответившие и о работе, и об оценках.

```python
rows = []
for wave in [3, 4]:
    data = df_long[df_long['wave'] == wave]
    marks_working = data.loc[data['works'] == 1, 'mark']
    marks_not_working = data.loc[data['works'] == 0, 'mark']

    result = mann_whitney_test(marks_working, marks_not_working)
    result['волна'] = wave
    rows.append(result)

h1_results = pd.DataFrame(rows).set_index('волна')
h1_results  # группа 1 — работающие, группа 2 — неработающие
```

**Вывод:** p-value = 0.57 на 3 курсе и 0.15 на 4 курсе, оба больше 0.05 — **H0 не отвергаю**. Медианы оценок у работающих и неработающих одинаковые («4 и 5»), средние почти совпадают (2.57 против 2.53 и 2.65 против 2.75). **Работа сама по себе не связана с успеваемостью.** Это согласуется с корреляционным анализом и со словами самих студентов: у 72% работающих успеваемость не страдает.

## Гипотеза 2. Работающие студенты чаще задумываются об отчислении?

Сравниваю долю студентов, думавших об отчислении (`dropout_thoughts` = 1), у работающих и неработающих. Это сравнение долей, поэтому использую хи-квадрат.

**H0:** доля задумывавшихся об отчислении одинаковая у работающих и неработающих.
**H1:** доли различаются.

Подвыборка: участники волны, ответившие и о работе, и о мыслях об отчислении. Дополнительно проверяю ту же гипотезу для четырёх типов занятости (`work_status`).

```python
rows = []
for wave in [3, 4]:
    data = df_long[df_long['wave'] == wave]
    for group_column in ['works', 'work_status']:
        result = chi2_test(data, group_column, 'dropout_thoughts')
        result['волна'] = wave
        result['группы'] = 'работает / нет' if group_column == 'works' else '4 типа занятости'
        rows.append(result)

h2_results = pd.DataFrame(rows).set_index(['волна', 'группы'])
h2_results
```

```python
# доли задумывавшихся об отчислении для наглядности
dropout_by_status = df_long.groupby(['wave', 'work_status'])['dropout_thoughts'].mean().unstack(0) * 100
dropout_by_status.loc[WORK_ORDER].round(1)
```

**Вывод:**
- **3 курс: H0 отвергаю** (p = 0.001). Работающие студенты чаще задумываются об отчислении: 47–53% против 40% у неработающих. Связь слабая (V Крамера = 0.09–0.10), но значимая.

- Условие хи-квадрат выполнено: минимальная ожидаемая частота — 22 и больше (нужно не меньше 5).
- Итог: на 3 курсе совмещение учёбы и работы связано с мыслями об отчислении. К 4 курсу различие сглаживается, возможно, потому что до 4 курса доходят те, кто научился совмещать.

## Гипотеза 3. Интенсивная работа сокращает сон?

В расчётах я увидел, что сон заметно короче только у студентов, работающих больше 30 часов в неделю. Проверяю это, сравнивая продолжительность сна в будни у группы «0 ч» и у каждой группы с работой. Продолжительность сна распределена не нормально, поэтому использую Манна–Уитни.

**H0:** продолжительность сна в будни у работающих N часов такая же, как у неработающих.
**H1:** продолжительность сна различается.

Подвыборка: участники волны, указавшие часы работы и время сна. Группа сравнения — «0 ч».

```python
rows = []
for wave in [3, 4]:
    data = df_long[df_long['wave'] == wave]
    sleep_no_work = data.loc[data['hours_group'] == '0 ч', 'sleep_weekday']
    for group_name in HOURS_ORDER[1:]:  # все группы, кроме «0 ч»
        sleep_group = data.loc[data['hours_group'] == group_name, 'sleep_weekday']
        result = mann_whitney_test(sleep_group, sleep_no_work)
        result['волна'] = wave
        result['часы работы'] = group_name
        rows.append(result)

h3_results = pd.DataFrame(rows).set_index(['волна', 'часы работы'])
h3_results[['n1', 'n2', 'среднее 1', 'среднее 2', 'p-value', 'H0']]  # 1 — группа с работой, 2 — «0 ч»
```

**Вывод:**
- **3 курс, больше 30 часов работы: H0 отвергаю** (p = 0.004). Такие студенты спят в среднем 7.0 часа против 7.7 у неработающих — на 40 минут меньше.
- При 11–30 часах работы разница в сне незначима.
- При 1–10 часах тест формально значим (p = 0.014), хотя у 11–30 часов разницы нет. Скорее всего, это случайность: я провёл 8 сравнений, и при таком числе проверок одна «ложная» значимость вполне ожидаема. С поправкой Бонферрони (α = 0.05 / 8 ≈ 0.006) значимым остаётся только сравнение «больше 30 часов».
- **4 курс:** ни одно различие не значимо. У работающих больше 30 часов сон всего на 15 минут короче.
- Итог: **сон страдает только при очень большой нагрузке (больше 30 часов в неделю) и только на 3 курсе.** Умеренная работа сон не сокращает.

## Гипотеза 4. Работа по специальности связана с лучшей учёбой?

Использую работающих студентов 4 курса (`workers_4w`) с колонкой `major_match` из раздела расчётов. Проверяю три связи (везде хи-квадрат, так как сравниваю доли):
1. работа по специальности и **хорошие оценки**;
2. работа по специальности и согласие с тем, что **работа помогает учиться**;
3. работа по специальности и согласие с тем, что **страдает успеваемость**.

**H0:** доли одинаковые у работающих по специальности и не по специальности.
**H1:** доли различаются.

Подвыборка: работающие студенты 4 курса, ответившие на вопрос о соответствии работы специальности и на соответствующий вопрос. Тех, кто затруднился с ответом, в выборке нет: служебные коды я заменил на пропуски в предобработке.

```python
# флаги для работающих студентов 4 курса: 1 — да, 0 — нет
workers_4w['good_marks'] = make_flag(workers_4w['mark_4w'], [3, 4])
workers_4w['work_helps'] = make_flag(workers_4w['work5_4w_1'], [3, 4])  # «скорее» или «полностью согласен»
workers_4w['marks_suffer'] = make_flag(workers_4w['work5_4w_2'], [3, 4])
workers_4w['hard_to_combine'] = make_flag(workers_4w['work5_4w_3'], [3, 4])

h4_checks = {
    'good_marks': 'Хорошие оценки',
    'work_helps': 'Работа помогает учиться',
    'marks_suffer': 'Страдает успеваемость',
}

rows = []
for flag, label in h4_checks.items():
    result = chi2_test(workers_4w, 'major_match', flag)
    result['показатель'] = label
    result['по специальности, %'] = round(workers_4w.loc[workers_4w['major_match'] == 'По специальности', flag].mean() * 100, 1)
    result['не по специальности, %'] = round(workers_4w.loc[workers_4w['major_match'] == 'Не по специальности', flag].mean() * 100, 1)
    rows.append(result)

h4_results = pd.DataFrame(rows).set_index('показатель')
h4_results
```

**Вывод:**
- **Хорошие оценки: H0 отвергаю** (p = 0.02). У работающих по специальности хорошие оценки у 63%, у работающих не по специальности — у 51%. Связь слабая (V = 0.12).
- **«Работа помогает учиться»: H0 отвергаю** (p < 0.001). 77% против 36% — это самая сильная связь в проекте (V Крамера = 0.41, средняя сила).
- **«Страдает успеваемость»: H0 не отвергаю** (p = 1.0). Доли одинаковые — 27%.
- Итог: работа по специальности **даёт пользу для учёбы и связана с лучшими оценками, но не снижает нагрузку**: жалобы на успеваемость встречаются одинаково часто. Условие хи-квадрат выполнено (ожидаемые частоты от 54).

Сравниваю работающих студентов 4 курса с разными мотивами (`motive`: ради денег / ради развития) по трём оценкам совмещения. Метод — хи-квадрат.

**H0:** доля согласных с утверждением одинаковая у работающих ради денег и ради развития.
**H1:** доли различаются.

Подвыборка: работающие студенты 4 курса, назвавшие мотив (кроме варианта «другое») и ответившие на утверждение.

```python
h5_checks = {
    'hard_to_combine': 'Трудно совмещать',
    'marks_suffer': 'Страдает успеваемость',
    'work_helps': 'Работа помогает учиться',
    'good_marks': 'Хорошие оценки',
}

rows = []
for flag, label in h5_checks.items():
    result = chi2_test(workers_4w, 'motive', flag)
    result['показатель'] = label
    result['ради денег, %'] = round(workers_4w.loc[workers_4w['motive'] == 'Ради денег', flag].mean() * 100, 1)
    result['ради развития, %'] = round(workers_4w.loc[workers_4w['motive'] == 'Ради развития', flag].mean() * 100, 1)
    rows.append(result)

h5_results = pd.DataFrame(rows).set_index('показатель')
h5_results
```

## Гипотеза 6. Начало работы ухудшает оценки и сон у одного и того же студента?

Использую таблицу `panel` из раздела расчётов: студенты обеих волн с траекторией занятости. Для каждой траектории сравниваю показатели одного и того же студента на 3 и на 4 курсе. Значения связанные (один человек дважды), поэтому использую **критерий Уилкоксона** для связанных выборок.

**H0:** оценки (сон) студента на 3 и 4 курсе не различаются.
**H1:** оценки (сон) изменились.

Подвыборка: студенты обеих волн, ответившие о работе, а для каждого показателя — ответившие на этот вопрос в обеих волнах. Оценки беру в исходной шкале 1–4, а не флагом: так видно и небольшие изменения.

```python
# добавляю в panel оценки по шкале 1–4 из df_long
marks_3w = df_long.loc[df_long['wave'] == 3, ['student_id', 'mark']].rename(columns={'mark': 'mark_3w'})
marks_4w = df_long.loc[df_long['wave'] == 4, ['student_id', 'mark']].rename(columns={'mark': 'mark_4w'})
panel = panel.merge(marks_3w, on='student_id').merge(marks_4w, on='student_id')

rows = []
for name in TRAJECTORY_ORDER:
    group = panel[panel['trajectory'] == name]
    for measure, label in [('mark', 'Оценки (1–4)'), ('sleep_weekday', 'Сон в будни, ч')]:
        pair = group[[f'{measure}_3w', f'{measure}_4w']].dropna().astype('float')  # ответили в обеих волнах
        statistic, p_value = st.wilcoxon(pair[f'{measure}_3w'], pair[f'{measure}_4w'])
        rows.append({
            'траектория': name,
            'показатель': label,
            'n': len(pair),
            'среднее 3 курс': round(pair[f'{measure}_3w'].mean(), 2),
            'среднее 4 курс': round(pair[f'{measure}_4w'].mean(), 2),
            'p-value': round(p_value, 4),
            'H0': 'отвергаю' if p_value < 0.05 else 'не отвергаю',
        })

h6_results = pd.DataFrame(rows).set_index(['показатель', 'траектория'])
h6_results
```

Что получили: в `panel` добавились оценки по шкале 1–4 на 3 и 4 курсе (`mark_3w`, `mark_4w`), а в `h6_results` — результаты теста Уилкоксона для каждой траектории.

Изменение у тех, кто **начал работать**, само по себе ничего не доказывает: у всех студентов к 4 курсу может что-то меняться. Поэтому дополнительно сравниваю **величину изменения** у тех, кто начал работать, и у тех, кто не работал оба года (Манн–Уитни для разностей «4 курс − 3 курс»).

```python
panel['mark_change'] = panel['mark_4w'].astype('float') - panel['mark_3w'].astype('float')
panel['sleep_change'] = panel['sleep_weekday_4w'] - panel['sleep_weekday_3w']

started = panel[panel['trajectory'] == 'Начал работать']
never = panel[panel['trajectory'] == 'Не работал -> не работает']
always = panel[panel['trajectory'] == 'Работал -> работает']

rows = []
for column, label in [('mark_change', 'Изменение оценок'), ('sleep_change', 'Изменение сна, ч')]:
    for compared_name, compared in [('Начал работать', started), ('Работал -> работает', always)]:
        result = mann_whitney_test(compared[column], never[column])
        result['показатель'] = label
        result['группа 1'] = compared_name
        rows.append(result)

h6_changes = pd.DataFrame(rows).set_index(['показатель', 'группа 1'])
h6_changes[['n1', 'n2', 'среднее 1', 'среднее 2', 'p-value', 'H0']]  # группа 2 — не работал оба года
```

**Вывод:**
- **Внутри траекторий:** у студентов, которые начали работать, оценки к 4 курсу не изменились значимо (p = 0.09), а сон даже вырос (с 7.6 до 8.0 часа, p = 0.015). У тех, кто не работал оба года, оценки и сон тоже выросли (p < 0.05). Значит, рост — общая тенденция 4 курса, а не эффект работы.
- **Сравнение изменений:** изменения у тех, кто начал работать, **не отличаются** от изменений у тех, кто не работал (оценки p = 0.80, сон p = 0.69). У тех, кто работал оба года, тоже нет значимых отличий (p = 0.43 и 0.19).
- Итог: **на одних и тех же студентах не видно, что начало работы ухудшает оценки или сон.** Это важный аргумент: в лонгитюде я сравниваю человека с самим собой, поэтому различия между «сильными» и «слабыми» студентами не мешают выводу.

## Гипотеза 7. Причину «трудности совмещения учёбы и работы» чаще называют работающие полный день?

Среди работающих студентов, которые задумывались об отчислении, сравниваю, как часто они называют причину «трудности с совмещением учёбы и работы» (`dropout2_7`) при разных типах занятости. Метод — хи-квадрат.

**H0:** доля называющих эту причину одинаковая при любом типе занятости.
**H1:** доля зависит от типа занятости.

Подвыборка: только **работающие** студенты, которые **задумывались об отчислении**. Неработающих не беру: у них этой причины быть не может, они бы только «размыли» сравнение. Волны объединяю, иначе группа «полный день» слишком маленькая для хи-квадрат. При этом студент из обеих волн может попасть в выборку дважды — это ограничение проверки.

```python
working_with_thoughts = df_long[(df_long['works'] == 1) & (df_long['dropout_thoughts'] == 1)]
print(f'Работающих, задумывавшихся об отчислении: {len(working_with_thoughts)}')

# доля называющих причину «совмещение учёбы и работы» в каждой группе, %
h7_shares = working_with_thoughts.groupby('work_status')['dropout2_7'].mean() * 100
print(h7_shares.loc[WORK_ORDER[:3]].round(1).to_string())

h7_result = chi2_test(working_with_thoughts, 'work_status', 'dropout2_7')
pd.Series(h7_result, name='результат').to_frame()
```

**Вывод:** p-value = 0.0005 — **H0 отвергаю**. Среди работающих студентов, задумывавшихся об отчислении, причину «трудности совмещения учёбы и работы» называют **61% работающих полный день, 44% работающих неполный день и 35% с приработками**. Связь слабая–средняя (V Крамера = 0.18), условие хи-квадрат выполнено (ожидаемые частоты от 30). Чем интенсивнее занятость, тем чаще именно работа становится причиной мыслей об отчислении.

# Построение регрессии

В гипотезах я сравнивал группы по одному признаку за раз. Но признаки связаны между собой: например, работающие могут чаще учиться платно или чаще быть из определённого направления. Регрессия позволяет оценить связь работы с результатами, **учитывая остальные характеристики студента одновременно**.

**Какую регрессию строю.** Главные результаты у меня бинарные (хорошие оценки — да/нет, думал об отчислении — да/нет). Обычная линейная регрессия для них не подходит: она может предсказать «вероятность» меньше 0 или больше 1. Поэтому использую **логистическую регрессию** (`smf.logit` из statsmodels). Для продолжительности сна — количественной переменной — использую обычную линейную регрессию (`smf.ols`).

**Как читать результат логистической регрессии — отношение шансов (OR):**
- OR = 1 — признак не связан с результатом;
- OR > 1 — шансы результата выше (например, OR = 1.5 — шансы в 1.5 раза выше, чем у группы сравнения);
- OR < 1 — шансы ниже;
- если 95% доверительный интервал OR включает 1, связь незначима.

**Модели:**
1. Хорошие оценки ~ тип занятости + контрольные признаки (обе волны).
2. Мысли об отчислении ~ тип занятости + контрольные признаки (обе волны).
3. Сон в будни ~ часы работы + контрольные признаки (обе волны).
4. Только работающие студенты 4 курса: хорошие оценки и «работа помогает учиться» ~ соответствие специальности + мотив + часы работы + контроли.

**Контрольные признаки:** волна, пол, форма обучения, направление подготовки, материальное положение семьи.

**Один студент — две строки.** Модели 1–3 строю по `df_long`, где студенты из обеих волн встречаются дважды. Их ответы не независимы, поэтому стандартные ошибки считаю **с кластеризацией по студенту** (`cov_type='cluster'`): так p-value не будут заниженными.

## Подготовка признаков для регрессии

В контрольных признаках есть пропуски: например, направление подготовки не указано у части студентов. Если удалить таких студентов, потеряется много данных. Поэтому пропуск делаю **отдельной категорией** («Не указано»). Мелкие направления (сельское хозяйство, искусство — единицы студентов) объединяю в «Другое», а материальное положение семьи — в три группы.

В каждой категориальной переменной выбираю **группу сравнения** — с ней сравниваются остальные группы:
- тип занятости — «Не работает»;
- направление — «Инженерные» (самое многочисленное);
- форма обучения — «Бюджет»;
- материальное положение — «Среднее»;
- пол — «Мужской»;
- волна — 3.

```python
major_names = {1: 'Естественные науки', 2: 'Инженерные', 3: 'Другое', 4: 'Другое',
               5: 'Науки об обществе', 6: 'Другое', 7: 'Гуманитарные', 8: 'Другое'}
df_long['major'] = df_long['bigmajor'].map(major_names).fillna('Не указано')
df_long['sex_label'] = df_long['sex'].map({1: 'Мужской', 2: 'Женский'}).fillna('Не указан')
df_long['funding_label'] = df_long['funding'].map({1: 'Бюджет', 2: 'Платное', 3: 'Целевое'}).fillna('Не указано')


def ses_group(code):
    """Материальное положение семьи: 1–3 — низкое, 4–5 — среднее, 6 — высокое."""
    if pd.isna(code):
        return 'Не указано'
    if code <= 3:
        return 'Низкое'
    if code <= 5:
        return 'Среднее'
    return 'Высокое'


df_long['ses_group'] = df_long['ses1'].apply(ses_group)

control_columns = ['wave', 'sex_label', 'funding_label', 'major', 'ses_group']
for column in control_columns:
    print(df_long[column].value_counts().to_dict())
```

Что получили: в `df_long` четыре новые колонки с контрольными признаками (`major`, `sex_label`, `funding_label`, `ses_group`), без пропусков.

Пишу формулу контрольных признаков один раз и функцию, которая превращает результат модели в понятную таблицу отношений шансов.

```python
# формула контрольных признаков; Treatment('...') задаёт группу сравнения
controls = (
    "C(wave) + C(sex_label, Treatment('Мужской')) + C(funding_label, Treatment('Бюджет'))"
    " + C(major, Treatment('Инженерные')) + C(ses_group, Treatment('Среднее'))"
)


def clean_name(name):
    """Делает название коэффициента читаемым: C(major, Treatment('Инженерные'))[T.Гуманитарные] -> major: Гуманитарные."""
    if not name.startswith('C('):
        return name
    variable = name[2:].split(',')[0].split(')')[0]
    level = name.split('[T.')[1].rstrip(']')
    return f'{variable}: {level}'


def odds_ratio_table(model):
    """Таблица отношений шансов: OR, 95% доверительный интервал и p-value для каждого признака."""
    confidence = model.conf_int()
    table = pd.DataFrame({
        'OR': np.exp(model.params),
        'OR нижняя граница': np.exp(confidence[0]),
        'OR верхняя граница': np.exp(confidence[1]),
        'p-value': model.pvalues,
    })
    table = table.drop(index='Intercept')  # константа не интерпретируется
    table.index = [clean_name(name) for name in table.index]
    table['значим'] = np.where(table['p-value'] < 0.05, 'да', 'нет')
    return table.round(3)


def fit_logit(formula, data, outcome):
    """Логистическая регрессия с кластеризацией ошибок по студенту. Строки с пропусками в результате убираются."""
    model_data = data.dropna(subset=[outcome, 'work_status'])
    model = smf.logit(formula, data=model_data).fit(
        cov_type='cluster', cov_kwds={'groups': model_data['student_id']}, disp=False)
    print(f'Наблюдений: {int(model.nobs)}, псевдо-R² Макфаддена: {model.prsquared:.3f}')
    return model
```

Что получили:
- `controls` — часть формулы с контрольными признаками, которую я добавляю в каждую модель;
- `clean_name()` — переводит технические названия коэффициентов в читаемые;
- `odds_ratio_table()` — таблица отношений шансов по модели;
- `fit_logit()` — строит логистическую регрессию с кластеризацией по студенту и печатает число наблюдений и качество модели.

**Псевдо-R² Макфаддена** — аналог R² для логистической регрессии. Значения 0.2–0.4 считаются очень хорошими, но для опросных данных о поведении людей обычны значения меньше 0.05: на оценки влияет множество факторов, которых нет в данных (способности, мотивация, сложность программы).

**Правило «10 наблюдений на признак».** В моделях 1–3 около 15 признаков (с учётом категорий) и больше 1500 наблюдений — данных хватает с большим запасом.

## Модель 1. Хорошие оценки

```python
work_term = "C(work_status, Treatment('Не работает'))"
model_marks = fit_logit(f'good_marks ~ {work_term} + {controls}', df_long, 'good_marks')
marks_or = odds_ratio_table(model_marks)
marks_or
```

**Вывод** (2043 наблюдения, псевдо-R² = 0.05):
- **Тип занятости при учёте остальных признаков с оценками значимо не связан.** У работающих неполный день OR = 1.15, у работающих с приработками — 0.88, оба незначимы. У работающих полный день шансы на хорошие оценки ниже на 31% (OR = 0.69), но p = 0.08, то есть на уровне 0.05 связь незначима.
- Гораздо сильнее с оценками связаны **пол и направление**: у девушек шансы на хорошие оценки в 1.7 раза выше, у студентов наук об обществе и гуманитарных направлений — в 2.1–2.5 раза выше, чем у инженеров. На 4 курсе шансы выше, чем на 3-м (OR = 1.3).
- Итог: работа — не главный фактор успеваемости. Полная занятость, возможно, немного снижает шансы на хорошие оценки, но этот эффект на грани значимости.

## Модель 2. Мысли об отчислении

```python
model_dropout = fit_logit(f'dropout_thoughts ~ {work_term} + {controls}', df_long, 'dropout_thoughts')
dropout_or = odds_ratio_table(model_dropout)
dropout_or
```

**Вывод** (2125 наблюдений, псевдо-R² = 0.02):
- **Работа на полный день значимо повышает шансы задумываться об отчислении — в 1.7 раза** (OR = 1.74, p = 0.004). Это самый сильный признак в модели.
- **Нерегулярные приработки** тоже связаны с мыслями об отчислении (OR = 1.35, p = 0.01), а **работа на неполный день — нет** (OR = 1.18, p = 0.16). Возможно, приработки чаще бывают вынужденными (нужны деньги) и менее предсказуемыми по нагрузке.
- Из контрольных признаков значимы низкое материальное положение семьи (OR = 1.73) и женский пол (OR = 1.38). На 4 курсе шансы ниже (OR = 0.81).
- Итог: связь полной занятости с мыслями об отчислении **сохраняется**, даже когда учтены пол, направление, форма обучения и материальное положение.

Сравниваю отношения шансов для типа занятости в двух моделях на одном графике. Точка — OR, линия — 95% доверительный интервал. Если линия пересекает вертикальную черту OR = 1, связь незначима.

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 4), sharey=True)
for ax, table, title in [(axes[0], marks_or, 'Модель 1: хорошие оценки'),
                         (axes[1], dropout_or, 'Модель 2: мысли об отчислении')]:
    work_rows = table[table.index.str.startswith('work_status')]  # только тип занятости
    labels = [name.replace('work_status: ', '') for name in work_rows.index]
    errors = [work_rows['OR'] - work_rows['OR нижняя граница'], work_rows['OR верхняя граница'] - work_rows['OR']]
    ax.errorbar(work_rows['OR'], labels, xerr=errors, fmt='o', capsize=5, color='#2a6f97')
    ax.axvline(1, color='grey', linestyle='--')
    ax.set_title(title)
    ax.set_xlabel('отношение шансов (группа сравнения — не работает)')
plt.tight_layout()
plt.show()
```

## Модель 3. Продолжительность сна

Сон в будни — количественная переменная, поэтому строю линейную регрессию (МНК). Вместо типа занятости беру группы по часам работы (`hours_group`), группа сравнения — «0 ч». Коэффициент показывает, на сколько часов сон в группе отличается от сна неработающих при одинаковых остальных признаках.

Нормальность сна нарушена (тест Шапиро–Уилка выше), но при таком большом числе наблюдений оценки коэффициентов МНК остаются надёжными. Для p-value использую те же кластеризованные ошибки.

```python
sleep_data = df_long.dropna(subset=['sleep_weekday', 'hours_group'])
model_sleep = smf.ols(f"sleep_weekday ~ C(hours_group, Treatment('0 ч')) + {controls}", data=sleep_data).fit(
    cov_type='cluster', cov_kwds={'groups': sleep_data['student_id']})
print(f'Наблюдений: {int(model_sleep.nobs)}, R²: {model_sleep.rsquared:.3f}')

sleep_coefficients = pd.DataFrame({
    'коэффициент, ч': model_sleep.params,
    'p-value': model_sleep.pvalues,
}).drop(index='Intercept')
sleep_coefficients.index = [clean_name(name) for name in sleep_coefficients.index]
sleep_coefficients['в минутах'] = sleep_coefficients['коэффициент, ч'] * 60
sleep_coefficients.round(3)
```

**Вывод** (1991 наблюдение, R² = 0.025):
- **Больше 30 часов работы в неделю** — минус **28 минут сна** в будни по сравнению с неработающими (p = 0.008), даже с учётом остальных признаков.
- При 11–30 часах сон короче на 10 минут, это незначимо.
- При 1–10 часах сон значимо короче на 15 минут (p = 0.014). Тот же неожиданный результат был в гипотезе 3, и он не согласуется с отсутствием эффекта при 11–30 часах. Возможно, это нерегулярные приработки в вечернее время, но данных для проверки этого нет.
- Модель объясняет всего 2.5% разброса сна: работа — лишь один из многих факторов. Например, у студентов естественных наук сон короче на 25 минут, чем у инженеров.

## Модель 4. Тип работы (работающие студенты 4 курса)

Проверяю, остаётся ли связь работы по специальности и мотива с учёбой, если учесть **часы работы** и контрольные признаки. Например, вдруг по специальности работают просто меньше часов, и дело в нагрузке, а не в типе работы.

Подвыборка — работающие студенты 4 курса (`workers_4w`), ответившие на вопросы о соответствии специальности и мотиве. Студентов здесь около 400, поэтому контролей беру меньше: пол, форма обучения и направление. Один студент встречается один раз, кластеризация не нужна.

```python
workers_4w['major'] = workers_4w['bigmajor'].map(major_names).fillna('Не указано')
workers_4w['sex_label'] = workers_4w['sex'].map({1: 'Мужской', 2: 'Женский'}).fillna('Не указан')
workers_4w['funding_label'] = workers_4w['funding'].map({1: 'Бюджет', 2: 'Платное', 3: 'Целевое'}).fillna('Не указано')
workers_4w['hours_10'] = workers_4w['work_hours_4w'] / 10  # часы работы в десятках: коэффициент — на каждые 10 часов

type_formula = (
    "C(major_match, Treatment('Не по специальности')) + C(motive, Treatment('Ради денег')) + hours_10"
    " + C(sex_label, Treatment('Мужской')) + C(funding_label, Treatment('Бюджет')) + C(major, Treatment('Инженерные'))"
)

type_tables = {}
for outcome, title in [('good_marks', 'Хорошие оценки'), ('work_helps', 'Работа помогает учиться')]:
    model_data = workers_4w.dropna(subset=[outcome, 'major_match', 'motive', 'hours_10'])
    model = smf.logit(f'{outcome} ~ {type_formula}', data=model_data).fit(disp=False)
    print(f'{title}: наблюдений {int(model.nobs)}, псевдо-R² {model.prsquared:.3f}')
    table = odds_ratio_table(model)
    type_tables[title] = table.loc[['major_match: По специальности', 'motive: Ради развития', 'hours_10']]

pd.concat(type_tables, names=['результат', 'признак'])
```

Что получили: `type_tables` — словарь с двумя таблицами отношений шансов (для хороших оценок и для «работа помогает учиться»). Выведены только признаки работы: соответствие специальности, мотив и часы (на каждые 10 часов в неделю).

**Вывод:**
- **Работа по специальности остаётся значимой при учёте часов, мотива и контролей:**
  - шансы на хорошие оценки в 1.6 раза выше (OR = 1.63, p = 0.03);
  - шансы считать, что работа помогает учиться, в **6 раз выше** (OR = 6.2, p < 0.001).
- **Мотив** при учёте соответствия специальности становится незначимым (p = 0.10 и 0.19). То есть мотив «ради развития» связан с пользой для учёбы в основном потому, что такие студенты чаще работают по специальности.
- **Часы работы** не влияют ни на оценки, ни на пользу от работы (p = 0.67 и 0.31).
- Итог: для пользы от работы важна не нагрузка и не мотив, а **связь работы со специальностью**.

**Где совмещение приносит пользу или не вредит:**
1. **Успеваемость.** Оценки работающих не хуже, чем у неработающих. Это видно и при простом сравнении, и в регрессии с учётом пола, направления и материального положения.

4. **Работа не вытесняет спорт и внеучебную активность:** работающие занимаются спортом не реже, а время на клубы и курсы планируют такое же.

**Где совмещение создаёт трудности:**
1. **Интенсивность.** Проблемы появляются при **полной занятости и больше 30 часов в неделю**:
   - студенты 3 курса с такой нагрузкой спят на 30–40 минут меньше, а меньше 7 часов спят 38% из них против 20% у неработающих;
   - шансы задумываться об отчислении у работающих полный день в 1.7 раза выше;
   - 61% из них, задумываясь об отчислении, называют причиной именно трудности совмещения.
2. **Работа ради денег и не по специальности** чаще воспринимается как помеха: тем, кто работает ради денег, труднее совмещать (37% против 25%), и у них чаще страдает успеваемость (31% против 20%).
3. **Длительное совмещение.** Студенты, работавшие и на 3, и на 4 курсе, чаще задумываются об отчислении (около 50% против 35–37%).
4. **Нерегулярные приработки** тоже связаны с мыслями об отчислении (OR = 1.35), в отличие от стабильной работы на неполный день.

**Итог:** совмещение учёбы и работы само по себе не вредит. Результат зависит от **двух условий**:
- **сколько** студент работает: умеренная нагрузка (до ~20 часов в неделю) безопасна, полная занятость связана с трудностями;
- **где** он работает: работа по специальности приносит пользу для учёбы, работа «только ради денег» — чаще трудности.

**Рекомендации для университетов:**
- **Помогать студентам находить работу по специальности:** стажировки, карьерные центры, партнёрства с работодателями, засчитывание профильной работы в практику. Именно такая работа связана с пользой для учёбы.
- **Поддерживать студентов с высокой нагрузкой:** гибкое расписание, индивидуальные учебные планы, дистанционные форматы для работающих полный день. В этой группе выше риск мыслей об отчислении.
- **Учитывать работу как фактор риска отчисления:** «трудности совмещения» — четвёртая по частоте причина мыслей об отчислении. Её стоит отслеживать в мониторингах и учитывать в работе кураторов.
- **Финансовая поддержка** (стипендии, гранты) может снизить долю студентов, вынужденных работать ради денег, — а такая работа чаще создаёт трудности.

**Ограничения исследования:**
- **Связь — не причина.** Это опросные данные. Например, сильные студенты могут чаще находить работу по специальности, и тогда лучшие оценки — не следствие работы. Лонгитюдный анализ частично снимает эту проблему, но группы в нём небольшие.
- **Выборка** — селективные вузы, больше половины студентов — инженерные направления, большинство из обеспеченных семей. Для студентов массовых вузов результаты могут отличаться.
- **Осыпание выборки:** в 4-ю волну попали не все студенты 3-й, а около 600 анкет 4-й волны пришлось очистить. Отчислившиеся студенты в данные не попадают, поэтому негативные эффекты работы могут быть **недооценены**.
- **Самооценки:** оценки, часы работы и сон — со слов студентов, без проверки по документам.
- **Связи слабые:** работа объясняет небольшую часть различий между студентами. Успеваемость и мысли об отчислении зависят от многих факторов, которых нет в данных.
